# <span style="color:green">Introduction</span>


<p style="color: white; background-color: #1565C0; font-family: Verdana; padding: 10px; border-radius: 8px;">RTGS stands for Real Time Gross Settlement. It is operated by RBI and is used for large-value, instant fund transfers between banks. Unlike NEFT, RTGS settles transactions individually in real-time.</p>

# <span style="color:green">Description</span>

<p style="color: white; background-color: #1565C0; font-family: Verdana; padding: 10px; border-radius: 8px;">The dataset "Bank Wise RTGS Inward and Outward" sourced from the Reserve Bank of India provides a comprehensive breakdown of RTGS (Real-Time Gross Settlement) transactions conducted by different banks. The data spans both inward (incoming) and outward (outgoing) transactions and segregates them further based on interbank and customer-driven transactions.</p>


# <span style="color:green">Aim</span>

<p style="color: white; background-color: #1565C0; font-family: Verdana; padding: 10px; border-radius: 8px;">RTGS is the backbone of large-value transactions in India. This project aims to analyze how money flows INWARD and OUTWARD across different banks, identify banks with high transaction dominance, and treat extreme outliers that distort financial analysis. The analysis helps understand banking performance, market share, and risk.</p>


# <span style="color:green">Methodology</span>

<p style="color: white; background-color: #1565C0; font-family: Verdana; padding: 10px; border-radius: 8px;">The data is compiled by the Reserve Bank of India from the transaction reports submitted by individual banks on a regular basis.</p>

# <span style="color:green">Data insights</span>

<p style="color: white; background-color: #1565C0; font-family: Verdana; padding: 10px; border-radius: 8px;">From the dataset, several insightful narratives can be pieced together about the RTGS transaction landscape in India. One of the most pivotal points to consider is the transactional trend portrayed by individual banks. Observing the 'INWARD Total Volume' and 'OUTWARD Total Volume' provides a snapshot of which banks might be at the forefront of RTGS transactions. Those with consistently high volumes might be perceived as the more dominant or trusted entities in this sphere. Diving deeper, the dataset does a commendable job differentiating between interbank and customer transactions. Banks showing a substantial INWARD Volume Interbank' could be inferred as playing a more influential role in interbank settlements. This demarcation between interbank and customer-driven transactions offers a nuanced understanding of a bank's RTGS transactional ecosystem. However, the true depth of this dataset is realized when delving into the monetary values associated with these transactions. By juxtaposing INWARD Total Value' with 'INWARD Total Volume', one can decipher the average transaction size for each bank. This might unravel whether certain banks are more favored for large-scale corporate transactions or for smaller, perhaps retail-based transfers.</p>

```python
df['inward_interbank_amt'] = df['inward_interbank_amt'].fillna(df['inward_interbank_amt'].median())
df['inward_percentage_volume'] = df['inward_percentage_volume'].fillna(df['inward_percentage_volume'].median())
df['inward_costumer_amt'] = df['inward_costumer_amt'].fillna(df['inward_costumer_amt'].median())
df['inward_total_amt'] = df['inward_total_amt'].fillna(df['inward_total_amt'].median())
df['inward_percentage_amt'] = df['inward_percentage_amt'].fillna(df['inward_percentage_amt'].median())
df['outward_percentage_volume'] = df['outward_percentage_volume'].fillna(df['outward_percentage_volume'].median())
df['outward_interbank_amt'] = df['outward_interbank_amt'].fillna(df['outward_interbank_amt'].median())
```

```python
df['outward_costumer_amt'] = df['outward_costumer_amt'].fillna(df['outward_costumer_amt'].median())
df['outward_total_amt'] = df['outward_total_amt'].fillna(df['outward_total_amt'].median())
df['outward_percentage_amt'] = df['outward_percentage_amt'].fillna(df['outward_percentage_amt'].median())
```

```python
## changing data type of moth to date time
df['month'] = pd.to_datetime(df['month']) 
```

```python
## ADD 4 COLUMNS TO THE DATA SET

1.# Total RTGS Volume per bank per month
df['total_rtgs_volume'] = df['inward_total_volume'] + df['outward_total_volume']

2.# Total RTGS Amount per bank per month
df['total_rtgs_amt'] = df['inward_total_amt'] + df['outward_total_amt']

3.# Net Flow = Inward - Outward
df['net_flow_amt'] = df['inward_total_amt'] - df['outward_total_amt']
df['net_flow_volume'] = df['inward_total_volume'] - df['outward_total_volume']
4.# Year and Month for analysis
df['year'] = df['month'].dt.year
df['month_name'] = df['month'].dt.month_name()
```

```python
df.describe()
outliers_cols=[
    'inward_total_volume',
    'inward_total_amt',
    'outward_total_volume',
    'outward_total_amt'
]

for col in outliers_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit= Q1 - 1.5*IQR
    upper_limit = Q3 + 1.5*IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]
    print(col, ":", len(outliers), "outliers")
```

# <span style="color:gray"> 1.Importance of Finding Outliers </span>

<p style="color:black; background-color: #1565C0; font-family: Verdana; padding: 10px; border-radius: 8px;">

a) inward_total_volume & outward_total_volume: These columns show the total count of transactions. Finding outliers helps to identify which banks handle maximum number of RTGS transactions. Banks like SBI, HDFC, ICICI always appear as outliers because of their large customer base. It also helps to find unusual spikes in a particular month due to salary credit, festival season or year-end closing.

b) inward_total_amt & outward_total_amt: These columns show the total amount in Lakh INR. This is the most critical financial indicator. Finding outliers is important to understand market dominance, to detect high-value corporate and government transactions, and to identify potential data entry errors or money movement risks.
</p>

# <span style="color:gray"> 2. Capping of Outliers </span>

<p style="color:black; background-color: #1565C0; font-family: Verdana; padding: 10px; border-radius: 8px;">

Outliers in the four core columns - inward_total_volume, outward_total_volume, inward_total_amt and outward_total_amt - were handled using the capping method based on IQR. Instead of deleting outlier records of large banks like SBI and HDFC, their values were capped to the upper limit (Q3 + 1.5*IQR). This method retains all bank data without data loss, while controlling the extreme values. Capping ensures correct mean calculation, balanced visualizations.
</p>

```python
df[(df.inward_total_volume < lower_limit) | (df.inward_total_volume> upper_limit)]
```

```python
df[(df.inward_total_amt < lower_limit) | (df.inward_total_amt> upper_limit)]
```

```python
df[(df.outward_total_volume < lower_limit) | (df.outward_total_volume> upper_limit)]
```

```python
df[(df.outward_total_amt< lower_limit) | (df.outward_total_amt> upper_limit)]
```

```python
cols = ['inward_total_volume','outward_total_volume','inward_total_amt','outward_total_amt']

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    # Capping - limit the amount without removing values
    df[col] = df[col].clip(lower=lower, upper=upper)

    print(f"{col} -> Upper Limit: {upper}")
```

```python
# Before Capping
print("Before:", df[cols].mean())

# After Capping
for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3-Q1
    df[col] = df[col].clip(Q1-1.5*IQR, Q3+1.5*IQR)

print("After:", df[cols].mean())
```

### 1. Key Insights from RTGS Analysis

<div style="background-color:#e8f5e9; border-left:5px solid #2e7d32; padding:50px; border-radius:90px;">
  <b style="color:#2e7d32;"></b> 
A) Data Understanding:
Dataset contains bank-wise monthly RTGS transactions with inward/outward volume and amount. Data is spread across multiple years with 338+ banks, but highly dominated by top 6 banks.

B) Data Quality & Cleaning:
Around 10 columns had missing values in amount and percentage fields. All nulls were handled using median imputation to avoid distortion. No duplicate records found. Outliers were high in all 4 core columns - inward_total_volume, inward_total_amt, outward_total_volume, outward_total_amt due to presence of large banks like SBI. Capping using IQR method was applied to reduce skewness without data loss.

C) Feature Engineering:
New KPIs were created:
-total_rtgs_volume and total_rtgs_amt for overall transaction load.
-net_flow_amt and net_flow_volume to identify whether a bank is net receiver or sender.
-year, month_name for time trend analysis..
</div>
